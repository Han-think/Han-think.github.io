### S13. 월별 매출과 전월 대비 증감

```python
# TODO: 직접 쿼리를 작성하세요.
run_sql("""
SELECT *
FROM products
LIMIT 3;
""")
```

```python
# 예시 정답
run_sql("""
WITH monthly_sales AS (SELECT strftime('%Y-%m', o.orderDate) AS order_month, SUM(od.quantityOrdered * od.priceEach) AS sales FROM orders AS o JOIN orderdetails AS od ON o.orderNumber = od.orderNumber GROUP BY strftime('%Y-%m', o.orderDate)) SELECT order_month, ROUND(sales, 2) AS sales, ROUND(LAG(sales) OVER (ORDER BY order_month), 2) AS prev_sales, ROUND(sales - LAG(sales) OVER (ORDER BY order_month), 2) AS diff_sales FROM monthly_sales ORDER BY order_month;
""")
```
